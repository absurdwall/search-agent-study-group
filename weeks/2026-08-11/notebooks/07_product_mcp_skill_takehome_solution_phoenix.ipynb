#### Reveal the Skill solution

The button displays the complete Skill text but does not
execute or overwrite anything.

```python
import ipywidgets as widgets
from IPython.display import Code, display

solution_source = "---\nname: product-shopping-mcp\ndescription: Use when finding, inspecting, comparing, or recommending products through the remote product catalog.\nmetadata:\n  adk_additional_tools:\n    - execute\n---\n\n# Product shopping through MCP\n\nUse the activated Code Mode tool for grounded product discovery, inspection, comparison, and recommendations.\n\n## Procedure\n\n1. Call `execute` once with one sandboxed Python program for the whole product request.\n2. Inside that program, use `await call_tool(...)` to call `search_products` before retrieving details. Put the user's product type, size, feature, and budget constraints into one concise search query.\n3. Select exactly two qualifying product IDs only from returned search rows. Enforce visible constraints and consider both rating and review count when the user asks for highly rated products.\n4. Keep each returned product ID exactly as supplied, then call `product_details` twice—once for each selected ID.\n5. Return one compact object containing both full product records. After `execute` returns, compare only returned fields, clearly identify missing availability, seller, brand, or specification data, and recommend one candidate from the returned evidence.\n\n## Grounding rules\n\n- Never invent a product, product ID, price, rating, review count, seller, availability, brand, or specification.\n- Do not treat a search snippet as a complete product record or compare before both detail records return.\n- If search returns fewer qualifying candidates than requested, state how many were found.\n- If a tool returns an error, report it plainly and stop rather than guessing another ID or result.\n"
solution_language = 'markdown'
reveal_button = widgets.Button(description='I will sign up for 5K steps')
reveal_output = widgets.Output()


def reveal_solution(
    _,
    solution_source=solution_source,
    solution_language=solution_language,
    reveal_output=reveal_output,
):
    with reveal_output:
        reveal_output.clear_output()
        display(Code(data=solution_source, language=solution_language))


reveal_button.on_click(reveal_solution)
display(reveal_button, reveal_output)
```

### 🤖 Agent — assemble the small visible part

**You only need to change:** the four commented arguments inside
`Agent(...)`. Remove each leading `#`.

1. Keep the prepared name and model.
2. Use the supplied `PRODUCT_SYSTEM_INSTRUCTIONS` exactly as written.
3. Give the agent the prepared `product_skill_tools`.

The supplied system instruction says: use the product Skill for every
request; run one sandboxed program that searches first and retrieves
both full records; compare only returned fields; identify missing data;
and never invent product facts.

**You are done when:** the cell prints `🤖 Agent ready`.
If it fails, revisit the Agent cell and uncomment all four arguments.

```python
product_skill_tools = load_product_skill_tools(PRODUCT_SKILL_DIR)
product_agent = Agent(
    name="product_code_mode_agent",
    model=GOOGLE_MODEL,
    instruction=PRODUCT_SYSTEM_INSTRUCTIONS,
    tools=[product_skill_tools],
)
product_runner = make_product_runner(product_agent)
print("🤖 Agent ready.")
```

#### Reveal the Agent solution

```python
import ipywidgets as widgets
from IPython.display import Code, display

solution_source = 'product_skill_tools = load_product_skill_tools(PRODUCT_SKILL_DIR)\nproduct_agent = Agent(\n    name="product_code_mode_agent",\n    model=GOOGLE_MODEL,\n    instruction=PRODUCT_SYSTEM_INSTRUCTIONS,\n    tools=[product_skill_tools],\n)\nproduct_runner = make_product_runner(product_agent)\nprint("🤖 Agent ready.")'
solution_language = 'python'
reveal_button = widgets.Button(description='I will sign up for 5K steps')
reveal_output = widgets.Output()


def reveal_solution(
    _,
    solution_source=solution_source,
    solution_language=solution_language,
    reveal_output=reveal_output,
):
    with reveal_output:
        reveal_output.clear_output()
        display(Code(data=solution_source, language=solution_language))


reveal_button.on_click(reveal_solution)
display(reveal_button, reveal_output)
```
